```python
titanic = sns.load_dataset('titanic')
```

## 4. Plots

**Recreate the 7 plots below using the titanic dataframe. There are very few hints since most of the plots can be done with just one or two lines of code and a hint would basically give away the solution. Keep careful attention to the x and y labels for hints.**

```python
sns.jointplot(x='fare', y='age', data=titanic)
```

```python
sns.histplot(titanic['fare'], kde=False, bins=30, color='red')
```

```python
sns.boxplot(x='class', y='age', hue='class', data=titanic, dodge=False, legend=False)
```

```python
sns.swarmplot(x='class', y='age', hue='class', data=titanic, legend=False)
```

```python
sns.countplot(x='sex', hue='sex', data=titanic, legend=False)
```

```python
sns.heatmap(titanic.corr(numeric_only=True), cmap='coolwarm')
```

```python
g = sns.FacetGrid(titanic, col='sex')
g.map(plt.hist, 'age')
```
